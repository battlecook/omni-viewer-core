# Revenue analysis
Saved code and results for the first quarter.

Average revenue: $\bar{x}=\frac{1}{n}\sum_{i=1}^n x_i$.

| Month | Revenue |
|---|---:|
| January | 120 |
| February | 185 |
| March | 240 |




```python
import pandas as pd
revenue = pd.DataFrame({"month": ["Jan", "Feb", "Mar"], "revenue": [120, 185, 240]})
print("Loaded 3 rows")
revenue
```

```python
revenue.plot.bar(x="month", y="revenue")
```

```python
summary = {"total": int(revenue.revenue.sum()), "months": 3}
display(summary)
```
